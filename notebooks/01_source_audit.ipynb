# Source Audit

## Purpose

This notebook tests whether the NBA statistics data source can provide the
game-level information required for the 2026 NBA Playoffs project.

The goal is to retrieve a sample of playoff games, inspect the returned
columns, identify a known Knicks playoff game, and compare the API result
against NBA.com.

```python
from nba_api.stats.endpoints import leaguegamelog
import pandas as pd
```

```python
playoff_log = leaguegamelog.LeagueGameLog(
    season="2025-26",
    season_type_all_star="Playoffs"
)
```

```python
games = playoff_log.get_data_frames()[0]
```

## Sample Game Validation

Use Knicks vs. Hawks Game 1 on April 18, 2026 as the sample game.

The purpose of this check is to confirm that one NBA game produces two
team-game rows and that the API values match the official game result.

```python
sample_game_id = "0042500121"

sample_game = games[
    games["GAME_ID"] == sample_game_id
]

sample_game
```

```python
sample_game.shape
```

```python
sample_game[
    [
        "TEAM_ABBREVIATION",
        "MATCHUP",
        "WL",
        "FGM",
        "FGA",
        "FG3M",
        "FG3A",
        "FTM",
        "FTA",
        "OREB",
        "DREB",
        "AST",
        "TOV",
        "PTS",
        "PLUS_MINUS",
    ]
]
```

```python
sample_game[
    ["TEAM_ABBREVIATION", "WL", "PTS"]
]
```

```python
winner_rows = sample_game[sample_game["WL"] == "W"]
loser_rows = sample_game[sample_game["WL"] == "L"]

print("Winners:", len(winner_rows))
print("Losers:", len(loser_rows))
```

```python
winner_points = winner_rows["PTS"].iloc[0]
loser_points = loser_rows["PTS"].iloc[0]

print("Winner points:", winner_points)
print("Loser points:", loser_points)
print("Winner scored more:", winner_points > loser_points)
```

```python
unique_games = games["GAME_ID"].nunique()

print("Team-game rows:", len(games))
print("Unique games:", unique_games)
print("Expected team-game rows:", unique_games * 2)
```

```python
rows_per_game = games.groupby("GAME_ID").size()

rows_per_game.value_counts()
```

```python
wl_counts = (
    games.groupby(["GAME_ID", "WL"])
    .size()
    .unstack(fill_value=0)
)

wl_counts.head()
```

```python
valid_wl_structure = (
    (wl_counts["W"] == 1) &
    (wl_counts["L"] == 1)
).all()

print("Every game has exactly one winner and one loser:", valid_wl_structure)
```
